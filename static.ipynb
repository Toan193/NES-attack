```python
def visualize(clean_position=None, adv_position=None, gt=None, pred=None, name=None):
    fig, ax = plt.subplots(figsize=(5,5))
    if clean_position is not None:
        ax.plot(clean_position[:, 0], clean_position[:, 1], label='Past Trajectory', color='blue')
    if adv_position is not None:
        ax.plot(adv_position[:, 0], adv_position[:, 1], '-o', linewidth=1, markersize=2, label='Adversarial Trajectory', color='red')
    if pred is not None:
        if pred.ndim == 3:
            for i in range(pred.shape[0]):
                ax.plot(pred[i, :, 0], pred[i, :, 1], label='Prediction', color='yellow')
        else:
            ax.plot(pred[:, 0], pred[:, 1], label='Prediction', color='yellow')
    if gt is not None:
            ax.plot(gt[:, 0], gt[:, 1], label='Ground Truth', color='green')    

    ax.set_title('Vehicle Trajectory Visualization', fontsize=12, fontweight='bold')
    ax.set_xlabel('X Position (m)', fontsize=10)
    ax.set_ylabel('Y Position (m)', fontsize=10)
    ax.grid(True, linestyle='--', alpha=0.6)
    ax.legend(
        loc='best',
        fontsize=7,        
        labelspacing=0.2,    
        handletextpad=0.3,   
        borderpad=0.2,       
        handlelength=1.5     
    )

    ax.axis('equal')

    if name is not None:
        file_path = 'assets/' + name + '.png'
        plt.savefig(file_path)
        print('Đã lưu file vào ', file_path)

    plt.tight_layout()
    plt.show()

def to_numpy(string):
    list = string.apply(eval).values[0]
    array = np.array(list)
    return np.squeeze(array, axis=0)

def normalize_positions(positions, heading):
    t_current = positions[-1, :2]
    theta = heading[49]

    cos, sin = np.cos(theta), np.sin(theta)
    rot_mat = np.array([
        [cos, -sin],
        [sin, cos]
    ])

    local_positions = (positions - t_current) @ rot_mat
    return local_positions
```

```python
pos = data_results['clean_positions'][5]
heading = data_results['heading'][5]
position = normalize_positions(pos, heading)

noise = data_results['noise'][5]
adv_position = position + noise

gt = data_results['ground_truth'][5]
pred = data_results['prediction'][5]

visualize(position, adv_position, gt, pred)
```
